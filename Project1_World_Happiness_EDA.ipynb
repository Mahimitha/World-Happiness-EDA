```python
import pandas as pd

df = pd.read_csv('/content/2019.csv')

df.head()
```

```python
print("Shape:", df.shape)

print("\nColumns:")
print(df.columns)
```

```python
df.isnull().sum()
```

```python
df[['Country or region', 'Score']].sort_values(by='Score', ascending=False).head(10)
```

Q1: Which country was the happiest in 2019?                           

Ans: Based on the analysis, Finland was the happiest country in 2019 with a happiness score of 7.769. Denmark and Norway ranked second and third respectively. Most of the top 10 happiest countries are from Europe.

```python
df[['Country or region', 'Score']].sort_values(by='Score').head(10)
```

Q2: Which countries were the least happy in 2019?                     

Ans: South Sudan had the lowest happiness score in 2019, followed by Central African Republic and Afghanistan. Many of the least happy countries were affected by poverty, political instability, conflict, or poor living conditions.

```python
df['Score'].mean()
```

Q3: What is the average happiness score of all countries?          

Ans: The average happiness score across all countries is 5.407. Countries scoring above this value can be considered happier than average, while countries scoring below this value may have lower levels of well-being.

```python
df[['GDP per capita', 'Score']].corr()
```

Q4: Is GDP per capita related to happiness?                         

Ans: The correlation between GDP per capita and happiness score is 0.794, indicating a strong positive relationship. Countries with higher GDP per capita generally tend to have higher happiness scores.

```python
df[['Healthy life expectancy', 'Score']].corr()
```

Q5: Is healthy life expectancy related to happiness?               

Ans: The correlation between healthy life expectancy and happiness score is 0.780. This indicates that countries with higher life expectancy generally tend to have higher happiness levels.

```python
import matplotlib.pyplot as plt

top10 = df[['Country or region', 'Score']].sort_values(by='Score', ascending=False).head(10)

plt.figure(figsize=(10,5))
plt.bar(top10['Country or region'], top10['Score'])
plt.title('Top 10 Happiest Countries')
plt.xlabel('Country')
plt.ylabel('Happiness Score')
plt.xticks(rotation=45)
plt.show()
```

```python
top10 = df[['Country or region', 'Score']].sort_values(by='Score', ascending=False).head(10)

plt.figure(figsize=(10,5))
plt.plot(top10['Country or region'], top10['Score'], marker='o')
plt.title('Top 10 Happiest Countries')
plt.xlabel('Country')
plt.ylabel('Happiness Score')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df['Score'], bins=10)
plt.title('Distribution of Happiness Scores')
plt.xlabel('Happiness Score')
plt.ylabel('Number of Countries')
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df['GDP per capita'], df['Score'])
plt.title('GDP per Capita vs Happiness Score')
plt.xlabel('GDP per Capita')
plt.ylabel('Happiness Score')
plt.show()
```

```python
top5 = df[['Country or region', 'Score']].sort_values(by='Score', ascending=False).head(5)

plt.figure(figsize=(7,7))
plt.pie(top5['Score'], labels=top5['Country or region'], autopct='%1.1f%%')
plt.title('Top 5 Happiest Countries')
plt.show()
```

```python
import seaborn as sns

plt.figure(figsize=(8,6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

-> Insight 1
